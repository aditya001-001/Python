# File Handling in Python
Python provides built-in functions to perform read and write operations on files. The most common way is using the `open()` function.

## 1. Writing to a File
Modes:
- `w`: Write mode. Overwrites the file if it exists, creates it if it doesn't.
- `a`: Append mode. Appends data to the end of the file.

```python
# Open and write (standard way)
file = open("sample.txt", "w")
file.write("Hello! This is line 1.\n")
file.write("This is line 2.\n")
file.close() # Always close the file to free up system resources!

# Appending to the file
file = open("sample.txt", "a")
file.write("This line was appended later!\n")
file.close()
```

## 2. The `with` Statement (Context Manager)
The `with` statement simplifies file handling by automatically closing the file when the block inside is exited, even if an exception occurs.

```python
# Write using 'with'
with open("sample_with.txt", "w") as f:
    f.write("Writing using a context manager.\n")
    f.write("Automatic file closing included!\n")

# Read using 'with'
with open("sample_with.txt", "r") as f:
    content = f.read()
    print(content)
```

## 3. Reading Methods
- `read()`: Reads the entire file.
- `readline()`: Reads one line at a time.
- `readlines()`: Reads all lines and returns a list of strings.

```python
with open("sample.txt", "r") as f:
    # Read entire content
    print("--- Read entire file ---")
    f.seek(0) # reset file cursor to the beginning
    print(f.read())

    # Read line-by-line
    print("--- Read line-by-line ---")
    f.seek(0)
    print("Line 1:", f.readline().strip())
    print("Line 2:", f.readline().strip())

    # Read as list of lines
    print("--- Read as list of lines ---")
    f.seek(0)
    lines = f.readlines()
    print(lines)
```

## 4. Deleting a File
To delete a file, you must import the `os` module and use its `remove()` method.

```python
import os
if os.path.exists("sample.txt"):
    os.remove("sample.txt")
    print("sample.txt deleted successfully!")
else:
    print("File does not exist!")

if os.path.exists("sample_with.txt"):
    os.remove("sample_with.txt")
    print("sample_with.txt deleted successfully!")
```
