# Exception Handling in Python
An exception is an event that occurs during the execution of a program and disrupts the normal flow of instructions. Python uses `try`, `except`, `else`, and `finally` blocks to handle these runtime errors gracefully.

## 1. Basic Try-Except Block

```python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Error caught:", e)
```

## 2. Handling Multiple Exceptions
You can specify different except blocks to handle different exceptions.

```python
try:
    value = int(input("Enter a number: "))
    result = 100 / value
except ValueError:
    print("You must enter a valid integer!")
except ZeroDivisionError:
    print("Cannot divide by zero!")
```

## 3. The `else` and `finally` Blocks
- `else`: Runs if no exceptions were raised in the try block.
- `finally`: Runs ALWAYS, regardless of whether an exception occurred or not (used for cleanup).

```python
try:
    print("Opening a database connection...")
    num = 10
    denom = 2
    result = num / denom
except ZeroDivisionError:
    print("Error: Division by zero")
else:
    print(f"Division successful! Result = {result}")
finally:
    print("Closing connection (this runs no matter what).")
```

## 4. Raising Exceptions & Custom Exceptions
You can manually raise exceptions using the `raise` keyword, and define custom exception classes.

```python
# Defining custom exception
class InvalidAgeError(Exception):
    pass

def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    elif age < 18:
        raise InvalidAgeError("Must be 18 or older to register.")
    else:
        print("Registration successful!")

try:
    check_age(15)
except ValueError as e:
    print("ValueError caught:", e)
except InvalidAgeError as e:
    print("Custom Exception caught:", e)
```
