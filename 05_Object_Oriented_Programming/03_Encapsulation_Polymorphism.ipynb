# Encapsulation and Polymorphism

## 1. Encapsulation
Encapsulation restricts direct access to some of an object's components, preventing accidental modification. Python uses underscores to specify access level:
- `public_variable`: Accessible from anywhere.
- `_protected_variable`: A hint to not access directly (subclasses can use it).
- `__private_variable`: Restricts access using name mangling.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner          # Public
        self.__balance = balance    # Private attribute

    # Getter method
    def get_balance(self):
        return self.__balance

    # Setter method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New Balance: {self.__balance}")
        else:
            print("Invalid deposit amount!")

account = BankAccount("Aditya", 1000)
print("Account Owner:", account.owner)

# Direct access to __balance will fail
try:
    print(account.__balance)
except AttributeError as e:
    print("Cannot access __balance directly:", e)

# Accessing via Getter & Setter
print("Current Balance:", account.get_balance())
account.deposit(500)
```

## 2. Polymorphism
Polymorphism means "many forms". It allows different classes to have methods with the same name but different behaviors.

```python
class Cat:
    def make_sound(self):
        return "Meow"

class Dog:
    def make_sound(self):
        return "Woof"

# A common interface/function that accepts any object with a 'make_sound' method (Duck Typing)
def animal_sound_test(animal_object):
    print("Sound check:", animal_object.make_sound())

cat = Cat()
dog = Dog()

animal_sound_test(cat)
animal_sound_test(dog)
```
