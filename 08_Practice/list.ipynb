# Equilibrium index

```python
def equilibrium_index(arr):
    arr = [int(x) for x in arr]
    total=sum(arr)
    left_sum=0
    for i in range(len(arr)):
        right_sum=total-left_sum-arr[i]

        if right_sum==left_sum:
            return i
            break
        left_sum+=arr[i]



arr = ['1', '3', '5', '2', '2']
print(equilibrium_index(arr))
```
