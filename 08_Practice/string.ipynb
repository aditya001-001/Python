### B.1 Reverse a string without slicing.

```python
def rev(s):
    res=''
    for i in range(len(s)-1,-1,-1):
        res+=s[i]
    return res
print(rev("Aditya"))
```

### B.2 Check whether a string is a palindrome.

```python
def palindrome(s):
    rev=''
    for i in range(len(s)-1,-1,-1):
        rev+=s[i]
    if rev==s:
        return "Palindrome"
    else:
        return "Not a Palindrome"
print(palindrome('mom'))
print(palindrome('Aditya'))
```

### B.3 Count vowels and consonants.

```python
def vowel_conso_count(s):
    v=0
    c=0
    for i in s:
        if i in 'aeiouAEIOU':
            v+=1
        elif ('a' <= i <= 'z') or ('A' <= i <= 'Z'):
            c+=1
    return v,c
print(vowel_conso_count('Adityasdfsdglvhxbh'))
```

### B.4 Count the number of characters in a string without len().

```python
def lenght(s):
    count=0
    for i in s:
        count+=1
    return count
print(lenght('Adtiya670920r'))
```

### B.5 Count spaces in a string.

```python
def space_in_string(s):
    count=0
    for i in s:
        if i ==' ':
            count+=1
    return count
print(space_in_string("   a   "))
```
