```python
def room_required(n,starttime,endtime):
    def convert(time):
        h,m=map(int,time.split(':'))
        return h*60 + m
    starttime=[convert(x) for x in starttime]
    endtime=[convert(x) for x in endtime]
    starttime.sort()
    endtime.sort()
    i=0
    j=0
    rooms=0
    max_rooms=0
    while i<n:
        if starttime[i]<endtime[j]:
            rooms+=1
            max_rooms=max(max_rooms,rooms)
            i+=1
        else:
            rooms-=1
            j+=1

    return max_rooms
# try:
#     n = int(input("Enter the number of meetings: "))

#     starttime = []
#     endtime = []

#     for i in range(n):
#         start = input(f"Enter start time for meeting {i+1}: ")
#         end = input(f"Enter end time for meeting {i+1}: ")

#         starttime.append(start)
#         endtime.append(end)

#     print("Start times:", starttime)
#     print("End times:", endtime)

# except ValueError as e:
#     print("Invalid Entry:", e)
# print(room_required(n,starttime,endtime))
```
