```python
def exist(board, word):

    rows = len(board)
    cols = len(board[0])

    def dfs(i, j, index):

        # Word completely matched
        if index == len(word):
            return True

        # Outside matrix
        if i < 0 or i >= rows or j < 0 or j >= cols:
            return False

        # Character doesn't match
        if board[i][j] != word[index]:
            return False

        # Mark as visited
        temp = board[i][j]
        board[i][j] = '#'

        # Four directions
        directions = [(1,0), (-1,0), (0,1), (0,-1)]

        for di, dj in directions:
            if dfs(i + di, j + dj, index + 1):
                return True

        # Backtrack
        board[i][j] = temp

        return False

    # Try every cell as starting point
    for i in range(rows):
        for j in range(cols):
            if dfs(i, j, 0):
                return True

    return False


# -------- INPUT --------

# rows, cols = map(int, input("Enter Row and Column").split())

# board = []

# for _ in range(rows):
#     board.append(input(f"Enter the values in {_}row").split())

# word = input("Enter the Word").strip()

# -------- OUTPUT --------
word='ABBC'
board=[['A','B','A','C'],
       ['B','B','C','A'],
       ['A','A','A','D']]

if exist(board, word):
    print("True")
else:
    print("False")
```

```python
def matrix(board,s):
    row=len(board)
    col=len(board[0])
    def dfs(i,j,index):
        if index==len(s):
            return True
        if i<0 or i>=row or j<0 or j>=col:
            return False
        if board[i][j]!=s[index]:
            return False
        temp=board[i][j]
        board[i][j]='#'

        direc=[(1,0),(-1,0),(0,-1),(0,1)]


        for ki,mj in direc:
            if dfs(i+ki,j+mj,index+1):
                return True
        board[i][j]=temp
        return False

    for i in range(row):
        for j in range(col):
            if dfs(i,j,0):
                return True
    return False
print(matrix([['A','B','C'],
              ['A','B','C']],'ABBC'))
```

```python
def centered(s,n):
    remaining=n-len(s)
    left=remaining//2
    right=remaining-left
    return " "*left + s + " "*right
print(centered('Adi',30))
```

```python
def prime(n):
    cnt=0
    for i in range(1,n):
        if n%i==0:
            cnt+=1
    if cnt==1:
        return "Prime"
    else:
        cnt=0
        for i in range(2,n):
            if n%i==0:
                return i
print(prime(7))
print(prime(15))
print(prime(21))
print(prime(2))

```
