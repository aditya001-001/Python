# Creating a list

```python

numbers = [1, 2, 3, 4, 5]

print(numbers)
```

# Accessing elements

```python

fruits = ["apple", "banana", "mango"]

print(fruits[0])
print(fruits[1])
```

# Negative indexing

```python

print(fruits[-1])


# List slicing
print(numbers[1:4])
```

# append

```python

fruits.append("orange")

print(fruits)
```

# insert()

```python

fruits.insert(1, "grapes")

print(fruits)
```

# remove()

```python

fruits.remove("banana")

print(fruits)
```

# pop()

```python

fruits.pop(1)

print(fruits)
```

# Looping through list

```python

for item in fruits:
    print(item)


# Length of list
print(len(fruits))
```

# Sorting list

```python

numbers2 = [5, 2, 1, 4, 3]

numbers2.sort()

print(numbers2)
```

# List comprehension

```python

squares = [x * x for x in range(5)]

print(squares)
```
