#  STRING QUESTIONS   
## 🟢 Basic   
* Reverse a string without slicing.
* Check whether a string is a palindrome.
* Count vowels and consonants.
* Count the number of characters in a string without len().
* Count spaces in a string.
* Convert lowercase characters to uppercase without using .upper().
* Convert uppercase characters to lowercase without using .lower().
* Check whether a string contains only digits.
* Check whether a string contains only alphabets.
* Remove spaces from a string.
## 🟡 Intermediate  
* Count the frequency of each character.
* Find the most frequent character.
* Find the first non-repeating character.
* Find the first repeating character.
* Remove duplicate characters.
* Find the number of words in a sentence.
* Find the longest word in a sentence.
* Find the shortest word in a sentence.
* Reverse every word in a sentence.
* Reverse the order of words in a sentence.
## 🟠 Interview Level  
* Check whether two strings are anagrams.
* Find all duplicate characters.
* Find the character with the highest frequency.
* Check whether two strings are rotations of each other.
* Find the longest substring without repeating characters.
* Find the longest palindromic substring.
* Find all occurrences of a character in a string.
* Remove all occurrences of a given character.
* Find the number of occurrences of one string inside another.
* Check whether one string is a substring of another without using in.
