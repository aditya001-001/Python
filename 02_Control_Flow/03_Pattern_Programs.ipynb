# Controlling the number of columns

```python
row = int(input("Rows:"))
col = int(input("Col:"))
for i in range(row):
    for j in range(col):
        print("*",end=" ")
    print()
```

# "*" Triangle/Numbers Patterns


```python
row=int(input("row:"))
col=int(input("col:"))
for i in range(row):
    for j in range(col):
        print(f"[{i},{j}]",end="")
    print()
```

```python
n=int(input("N:"))
for i in range(n):
    for j in range(n):
        if  j<=i:
            print("*",end=" ")
        else:
            print(" " ,end=" ")
    print()
```
