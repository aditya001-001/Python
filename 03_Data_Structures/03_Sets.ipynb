# Python Sets
A set is an unordered, unindexed, and mutable collection of unique elements. Sets are written with curly brackets `{}`.

```python
# Creating a set
my_set = {1, 2, 3, 3, 4, 4, 5}
print("Set (duplicates removed):", my_set)

# Note: To create an empty set, use set() not {}
empty_dict = {}
empty_set = set()
print("Type of {}:", type(empty_dict))
print("Type of set():", type(empty_set))
```

## Adding and Removing Elements

```python
colors = {"red", "green", "blue"}

# Adding elements
colors.add("yellow")
print("After adding:", colors)

# Removing elements
colors.remove("green")  # raises KeyError if not found
colors.discard("purple") # does NOT raise error if not found
print("After removing/discarding:", colors)
```

## Set Operations
Sets support mathematical operations like union, intersection, difference, and symmetric difference.

```python
setA = {1, 2, 3, 4, 5}
setB = {4, 5, 6, 7, 8}

# Union
print("Union:", setA | setB)  # or setA.union(setB)

# Intersection
print("Intersection:", setA & setB)  # or setA.intersection(setB)

# Difference (A - B)
print("Difference (A-B):", setA - setB)  # elements in A but not in B

# Symmetric Difference
print("Symmetric Difference:", setA ^ setB)  # elements in either A or B, but not both
```
