# Python Tuples
A tuple is an ordered, immutable collection of elements. Tuples allow duplicate values and are defined using parentheses `()`.

```python
# Creating a tuple
my_tuple = (1, 2, 3, 2, 4)
print("Tuple:", my_tuple)
print("Type:", type(my_tuple))

# Single element tuple (Requires a trailing comma!)
single_element = (5,)
print("Single element tuple:", single_element)
print("Not a tuple (without comma):", type((5)))
```

## Accessing elements and Slicing

```python
fruits = ("apple", "banana", "cherry", "orange")
print("First item:", fruits[0])
print("Last item:", fruits[-1])
print("Slice:", fruits[1:3])
```

## Immutability
Once created, you cannot change, add, or remove items from a tuple.

```python
numbers = (1, 2, 3)
try:
    numbers[0] = 10
except TypeError as e:
    print("Error when modifying:", e)
```

## Tuple Packing and Unpacking
Assigning values to a tuple is packing. Extracting values back into variables is unpacking.

```python
# Packing
person = ("Aditya", 21, "Developer")

# Unpacking
name, age, profession = person
print("Name:", name)
print("Age:", age)
print("Profession:", profession)
```

## Tuple Methods
Tuples have only two built-in methods: `count()` and `index()`.

```python
nums = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5, 7)
print("Count of 7:", nums.count(7))
print("Index of 8 (first occurrence):", nums.index(8))
```
