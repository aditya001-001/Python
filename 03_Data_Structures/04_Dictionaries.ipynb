# Python Dictionaries
A dictionary stores data in key-value pairs. Dictionaries are mutable, and as of Python 3.7, they maintain insertion order. Keys must be unique and hashable.

```python
# Creating a dictionary
student = {
    "name": "Aditya",
    "age": 21,
    "courses": ["Python", "Machine Learning"],
    "is_graduated": False
}
print(student)
```

## Accessing Values

```python
# Accessing by Key
print("Name:", student["name"])

# Using .get() (Returns None or default if key doesn't exist, preventing KeyErrors)
print("Age:", student.get("age"))
print("GPA:", student.get("gpa", "Not Found"))
```

## Modifying and Adding Items

```python
# Modifying
student["age"] = 22

# Adding new key-value pair
student["gpa"] = 3.8
print("Modified Dictionary:", student)
```

## Dictionary Methods

```python
# Get all keys
print("Keys:", list(student.keys()))

# Get all values
print("Values:", list(student.values()))

# Get key-value tuples
print("Items:", list(student.items()))

# Iterating through dictionary
for key, val in student.items():
    print(f"{key} -> {val}")

# Removing items
gpa = student.pop("gpa")
print(f"Popped GPA: {gpa}")
print("Final Dictionary:", student)
```
