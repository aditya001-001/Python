# Database Connectivity in Python (SQLite)
Python comes with a built-in module called `sqlite3` to interact with SQL databases without needing external servers. This notebook walks through establishing a connection, creating a table, performing CRUD operations, and transaction management.

## 1. Establishing a Connection & Creating a Table
- Use `sqlite3.connect()` to open/create a database file.
- Use `.cursor()` to create a cursor object that executes SQL statements.

```python
import sqlite3

# Connect to database (creates file 'internship.db' if it doesn't exist)
conn = sqlite3.connect("internship.db")
cursor = conn.cursor()

# Create a table
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    course TEXT NOT NULL,
    grade REAL
)
""")
conn.commit()
print("Database connected and table created!")
```

## 2. Inserting Data (Secure Parameterization)
**Important**: Always use placeholders (`?`) instead of string formatting to prevent SQL injection attacks.

```python
# Securely inserting records
student_data = [("Aditya", "Python Basics", 95.5), ("Rahul", "Machine Learning", 88.0)]

cursor.executemany("""
INSERT INTO students (name, course, grade)
VALUES (?, ?, ?)
""", student_data)

conn.commit()
print(f"{cursor.rowcount} records inserted successfully!")
```

## 3. Querying / Reading Data
Use `.fetchall()` to retrieve all rows, or `.fetchone()` to retrieve one row at a time.

```python
cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()

print("--- Student Records ---")
for row in rows:
    print(f"ID: {row[0]} | Name: {row[1]} | Course: {row[2]} | Grade: {row[3]}")
```

## 4. Updating and Deleting Records

```python
# Update
cursor.execute("UPDATE students SET grade = ? WHERE name = ?", (98.0, "Aditya"))
conn.commit()
print("Record updated successfully!")

# Verify Update
cursor.execute("SELECT * FROM students WHERE name = ?", ("Aditya",))
print("Updated Record:", cursor.fetchone())

# Delete
cursor.execute("DELETE FROM students WHERE name = ?", ("Rahul",))
conn.commit()
print("Record deleted successfully!")
```

## 5. Safe Connection Closing using try-finally
Always close connections in a `finally` block to release file locks.

```python
try:
    # Show final remaining records
    cursor.execute("SELECT * FROM students")
    print("Final Students Table:", cursor.fetchall())
finally:
    # Clean up and close
    cursor.close()
    conn.close()
    print("Database connection closed safely.")

# Clean up the temporary database file
import os
if os.path.exists("internship.db"):
    os.remove("internship.db")
    print("Temporary database file removed.")
```
