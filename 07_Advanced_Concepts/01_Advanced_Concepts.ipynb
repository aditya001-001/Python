# Advanced Python Concepts
This notebook covers powerful Python features: Lambda functions, Map/Filter/Reduce, Decorators, and Generators.

## 1. Lambda Functions
Lambda functions are small, anonymous, one-line functions defined using the `lambda` keyword.

```python
# Regular function
def square(x):
    return x * x

# Lambda equivalent
square_lambda = lambda x: x * x

print("Square of 5 (regular):", square(5))
print("Square of 5 (lambda):", square_lambda(5))

# Lambda with multiple arguments
add = lambda a, b: a + b
print("Add 10 + 20:", add(10, 20))
```

## 2. Map, Filter, and Reduce
- `map()`: Applies a function to all items in an input list.
- `filter()`: Creates a list of elements for which a function returns True.
- `reduce()`: Performs a repetitive operation on sequential pairs of list elements (requires `functools`).

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

# 1. Map (square all numbers)
squares = list(map(lambda x: x * x, numbers))
print("Mapped (Squares):", squares)

# 2. Filter (get only even numbers)
evens = list(filter(lambda x: x % 2 == 0, numbers))
print("Filtered (Evens):", evens)

# 3. Reduce (calculate product of all elements)
product = reduce(lambda x, y: x * y, numbers)
print("Reduced (Product):", product)
```

## 3. Decorators
Decorators allow you to modify or extend the behavior of functions or methods without permanently modifying their code.

```python
# Defining a decorator
def my_decorator(func):
    def wrapper():
        print("[Log] Something is happening before the function is called.")
        func()
        print("[Log] Something is happening after the function is called.")
    return wrapper

# Applying the decorator
@my_decorator
def say_hello():
    print("Hello World!")

say_hello()
```

## 4. Generators
Generators are functions that return an iterator using the `yield` keyword. They yield values one at a time and suspend execution, saving massive amounts of memory compared to lists.

```python
# Generator function
def countdown(n):
    while n > 0:
        yield n
        n -= 1

# Using generator
gen = countdown(3)
print(next(gen))
print(next(gen))
print(next(gen))

# Using generator in a loop
for val in countdown(5):
    print(val, end=" ")
```
