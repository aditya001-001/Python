# Python Strings
Strings are sequences of characters enclosed in single, double, or triple quotes.

```python
# Creating Strings
single = 'Hello'
double = "World"
triple = '''This is a
multiline string'''

print(single)
print(double)
print(triple)
```

## String Indexing and Slicing
Python strings are indexed starting from 0. Negative indexing starts from -1 from the end.

```python
text = "Python Programming"

# Indexing
print("First char:", text[0])
print("Last char:", text[-1])

# Slicing: [start:stop:step]
print("Slice (0 to 6):", text[0:6])
print("Every second char:", text[::2])
print("Reversed string:", text[::-1])
```

## Immutability
Strings are immutable. Trying to change a character in a string raises a `TypeError`.

```python
text = "Hello"
try:
    text[0] = 'Y'  # Raises TypeError
except TypeError as e:
    print("Error:", e)
```

## String Formatting
Modern Python uses f-strings for string interpolation.

```python
name = "Aditya"
age = 21

# f-strings (Python 3.6+)
formatted = f"Hello, my name is {name} and I am {age} years old."
print(formatted)

# .format() method
formatted_old = "Name: {}, Age: {}".format(name, age)
print(formatted_old)
```

## Core String Methods

```python
text = "  Python is awesome!  "

print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Stripped:", text.strip())
print("Replace:", text.replace("awesome", "great"))
print("Split:", text.strip().split(" "))

words = ['Python', 'is', 'fun']
print("Join:", "-".join(words))
```
