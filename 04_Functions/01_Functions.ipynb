# Simple function

```python
def greet():
    print("Hello")

greet()
```

# Function with parameter

```python
def greet_user(name):
    print("Hello", name)

greet_user("Aditya")
```

# Function with return value

```python
def add(a, b):
    return a + b

result = add(10, 20)

print("Sum =", result)
```


# Default parameter

```python
def country(name="India"):
    print(name)

country()
country("Japan")
```

# Keyword arguments

```python

def student(name, age):
    print(name, age)

student(age=21, name="Aditya")
```

# *args example

```python
def numbers(*num):
    print(num)

numbers(1, 2, 3, 4)
```

# Recursive function

```python
def countdown(n):
    if n == 0:
        return

    print(n)

    countdown(n - 1)

countdown(5)
```
